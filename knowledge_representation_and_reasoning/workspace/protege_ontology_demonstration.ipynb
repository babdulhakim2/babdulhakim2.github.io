## 1. Ontology Class Hierarchy (15 Classes as Reported)

```python
def define_ontology_structure():
    """Define the complete ontology structure as implemented in Protégé"""
    
    # Class hierarchy (15 classes total)
    ontology_classes = {
        'Thing': {  # OWL root class
            'JobSeeker': {
                'ExperiencedJobSeeker': {},
                'EntryLevelJobSeeker': {}
            },
            'JobOpportunity': {
                'SeniorRole': {},
                'EntryLevelRole': {},
                'RemoteJob': {}
            },
            'Skill': {
                'TechnicalSkill': {},
                'SoftSkill': {}
            },
            'Organization': {},
            'Location': {},
            'Education': {},
            'Experience': {},
            'Requirement': {}
        }
    }
    
    # Object Properties (11 total)
    object_properties = [
        'hasSkill',
        'requiresSkill', 
        'hasExperience',
        'requiresExperience',
        'hasEducation',
        'requiresEducation',
        'worksAt',
        'offeredBy',
        'locatedIn',
        'appliesTo',
        'satisfies'
    ]
    
    # Data Properties (12 total)
    data_properties = [
        'experienceYears',
        'salaryExpectation',
        'salaryRange',
        'skillLevel',
        'jobTitle',
        'companyName',
        'locationName',
        'educationLevel',
        'yearsRequired',
        'jobDescription',
        'contactEmail',
        'postingDate'
    ]
    
    print("🏗️ Ontology Structure Definition")
    print("\n📋 CLASS HIERARCHY (15 classes):")
    
    def print_hierarchy(classes, indent=0):
        for class_name, subclasses in classes.items():
            print("  " * indent + f"• {class_name}")
            if subclasses:
                print_hierarchy(subclasses, indent + 1)
    
    print_hierarchy(ontology_classes)
    
    print(f"\n🔗 OBJECT PROPERTIES ({len(object_properties)} properties):")
    for i, prop in enumerate(object_properties, 1):
        print(f"   {i:2d}. {prop}")
    
    print(f"\n📊 DATA PROPERTIES ({len(data_properties)} properties):")
    for i, prop in enumerate(data_properties, 1):
        print(f"   {i:2d}. {prop}")
    
    return ontology_classes, object_properties, data_properties

# Define the structure
classes, obj_props, data_props = define_ontology_structure()
```

## 3. Reasoning Examples and DL Queries

```python
def demonstrate_reasoning_capabilities():
    """Demonstrate reasoning examples as would appear in Protégé"""
    
    print("🧠 Ontological Reasoning Demonstrations")
    print("=" * 45)
    
    # Automatic Classification Examples
    reasoning_examples = [
        {
            'input': 'AliceSmith hasExperience 7, hasSkill Python',
            'rule': 'hasExperience some integer[>= 5] → ExperiencedJobSeeker',
            'output': 'AliceSmith → ExperiencedJobSeeker ✓',
            'explanation': 'Automatic classification based on experience threshold'
        },
        {
            'input': 'SeniorDataScientistRole requiresExperience 5, locatedIn Manhattan',
            'rule': 'requiresExperience some integer[>= 4] → SeniorRole',
            'output': 'SeniorDataScientistRole → SeniorRole ✓',
            'explanation': 'Role classification based on experience requirements'
        },
        {
            'input': 'WebDeveloperPosition allowsRemoteWork true',
            'rule': 'allowsRemoteWork value true → RemoteJob',
            'output': 'WebDeveloperPosition → RemoteJob ✓',
            'explanation': 'Job type inference from work arrangement'
        }
    ]
    
    print("🔍 AUTOMATIC CLASSIFICATION EXAMPLES:")
    for i, example in enumerate(reasoning_examples, 1):
        print(f"\n   {i}. Input: {example['input']}")
        print(f"      Rule: {example['rule']}")
        print(f"      Result: {example['output']}")
        print(f"      Explanation: {example['explanation']}")
    
    # DL Query Examples
    dl_queries = [
        {
            'query': 'JobOpportunity and requiresSkill value Python',
            'description': 'Find all jobs requiring Python skills',
            'results': ['SeniorDataScientistRole', 'MLEngineerPosition'],
            'count': 2
        },
        {
            'query': 'JobSeeker and hasExperience some integer[>= 5]',
            'description': 'Find experienced job seekers (5+ years)',
            'results': ['AliceSmith', 'CarolWilson', 'DavidBrown'],
            'count': 3
        },
        {
            'query': 'SeniorRole and locatedIn value Manhattan',
            'description': 'Find senior roles in Manhattan',
            'results': ['SeniorDataScientistRole', 'TechLeadPosition'],
            'count': 2
        },
        {
            'query': 'JobSeeker and hasSkill value MachineLearning',
            'description': 'Find candidates with ML skills',
            'results': ['AliceSmith', 'EveJohnson'],
            'count': 2
        }
    ]
    
    print("\n🔎 DESCRIPTION LOGIC (DL) QUERIES:")
    for i, query in enumerate(dl_queries, 1):
        print(f"\n   {i}. Query: {query['query']}")
        print(f"      Description: {query['description']}")
        print(f"      Results: {', '.join(query['results'])} ({query['count']} total)")
    
    # Semantic Matching Example
    matching_scenario = {
        'candidate': 'AliceSmith',
        'candidate_profile': {
            'skills': ['Python', 'DataAnalysis', 'MachineLearning'],
            'experience': 7,
            'education': 'Masters'
        },
        'job': 'SeniorDataScientistRole',
        'job_requirements': {
            'skills': ['Python', 'MachineLearning', 'Statistics'],
            'experience': 5,
            'education': 'Bachelors'
        },
        'reasoning_chain': [
            'hasSkill(AliceSmith, Python) ∧ requiresSkill(SeniorDataScientistRole, Python) → skill_match',
            'hasSkill(AliceSmith, MachineLearning) ∧ requiresSkill(SeniorDataScientistRole, MachineLearning) → skill_match',
            'hasExperience(AliceSmith, 7) ∧ requiresExperience(SeniorDataScientistRole, 5) ∧ (7 ≥ 5) → experience_match',
            'hasEducation(AliceSmith, Masters) ∧ requiresEducation(SeniorDataScientistRole, Bachelors) → education_match'
        ],
        'match_score': 0.89,
        'explanation': 'Strong match: overlapping skills, exceeds experience requirement, higher education level'
    }
    
    print("\n🎯 SEMANTIC MATCHING EXAMPLE:")
    print(f"   Candidate: {matching_scenario['candidate']}")
    print(f"   Job: {matching_scenario['job']}")
    print(f"   \n   Reasoning Chain:")
    for step in matching_scenario['reasoning_chain']:
        print(f"     • {step}")
    print(f"   \n   Match Score: {matching_scenario['match_score']:.2f}")
    print(f"   Explanation: {matching_scenario['explanation']}")
    
    return reasoning_examples, dl_queries, matching_scenario

# Demonstrate reasoning
reasoning, queries, matching = demonstrate_reasoning_capabilities()
```

## 4. Ontology Structure Visualization

```python
def create_ontology_structure_diagram():
    """Create a visual representation of the ontology structure for IMAGE 2"""
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 10))
    fig.suptitle('Protégé Ontology Structure - Classes, Properties, and Relationships', 
                 fontsize=16, fontweight='bold')
    
    # Left side: Class Hierarchy
    G_classes = nx.DiGraph()
    
    # Add class hierarchy
    class_relationships = [
        ('Thing', 'JobSeeker'),
        ('Thing', 'JobOpportunity'),
        ('Thing', 'Skill'),
        ('Thing', 'Organization'),
        ('Thing', 'Location'),
        ('Thing', 'Education'),
        ('Thing', 'Experience'),
        ('Thing', 'Requirement'),
        ('JobSeeker', 'ExperiencedJobSeeker'),
        ('JobSeeker', 'EntryLevelJobSeeker'),
        ('JobOpportunity', 'SeniorRole'),
        ('JobOpportunity', 'EntryLevelRole'),
        ('JobOpportunity', 'RemoteJob'),
        ('Skill', 'TechnicalSkill'),
        ('Skill', 'SoftSkill')
    ]
    
    G_classes.add_edges_from(class_relationships)
    
    # Position classes hierarchically
    pos_classes = nx.spring_layout(G_classes, k=2, iterations=50)
    
    # Draw class hierarchy
    nx.draw(G_classes, pos_classes, ax=ax1, 
            with_labels=True, node_color='lightblue', 
            node_size=1500, font_size=8, font_weight='bold',
            arrows=True, arrowsize=10, edge_color='gray')
    
    ax1.set_title('A) Class Hierarchy (15 Classes)', fontweight='bold', pad=20)
    ax1.text(0.5, -0.1, 'IS-A relationships shown as arrows', 
             ha='center', transform=ax1.transAxes, fontsize=10, style='italic')
    
    # Right side: Property Relationships
    G_props = nx.Graph()
    
    # Add main entities and their relationships
    main_entities = ['JobSeeker', 'JobOpportunity', 'Skill', 'Organization', 'Education']
    G_props.add_nodes_from(main_entities)
    
    # Add property relationships
    property_relationships = [
        ('JobSeeker', 'Skill', 'hasSkill'),
        ('JobOpportunity', 'Skill', 'requiresSkill'),
        ('JobSeeker', 'Education', 'hasEducation'),
        ('JobOpportunity', 'Education', 'requiresEducation'),
        ('JobSeeker', 'JobOpportunity', 'appliesTo'),
        ('JobOpportunity', 'Organization', 'offeredBy')
    ]
    
    for source, target, prop in property_relationships:
        G_props.add_edge(source, target, label=prop)
    
    # Position properties
    pos_props = nx.spring_layout(G_props, k=3, iterations=50)
    
    # Draw property relationships
    nx.draw(G_props, pos_props, ax=ax2,
            with_labels=True, node_color='lightgreen',
            node_size=2000, font_size=9, font_weight='bold')
    
    # Add edge labels
    edge_labels = {(u, v): d['label'] for u, v, d in G_props.edges(data=True)}
    nx.draw_networkx_edge_labels(G_props, pos_props, edge_labels, ax=ax2, font_size=7)
    
    ax2.set_title('B) Object Properties (11 Properties)', fontweight='bold', pad=20)
    ax2.text(0.5, -0.1, 'Relationships between main classes', 
             ha='center', transform=ax2.transAxes, fontsize=10, style='italic')
    
    plt.tight_layout()
    plt.savefig('protege_ontology_structure.png', dpi=300, bbox_inches='tight')
    plt.show()
    
    print("📊 Ontology structure diagram saved as 'protege_ontology_structure.png'")
    print("   This visualization shows the class hierarchy and property relationships")
    print("   Perfect for IMAGE 2 in your report!")

# Create structure diagram
create_ontology_structure_diagram()
```
